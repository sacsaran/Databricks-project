# BRONZE LAYER

```python
%sql
CREATE OR REFRESH STREAMING TABLE bank_platform.bronze_banking.bronze_transaction 
AS 
SELECT *,
current_timestamp() as processing_ts,
_metadata.file_name as file_name,
_metadata.file_modification_time as file_modification_time

From STREAM read_files("/Volumes/bank_platform/bank_raw/landing/transactions/",
format => 'json',
header=>'True'
)
```

```python
%sql
CREATE OR REFRESH STREAMING TABLE bank_platform.bronze_banking.bronze_customers
AS 
SELECT *,
current_timestamp() as processing_ts,
_metadata.file_name as file_name,
_metadata.file_modification_time as file_modification_time
FROM STREAM read_files("/Volumes/bank_platform/bank_raw/landing/customers_cdc/",
FORMAT => 'json',
header => 'true'
)
```
