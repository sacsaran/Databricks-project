# CREATING SILVER

```python

spark.sql(

"""
SELECT 
        to_date(date,'dd/MM/yyyy') as Date,
        year(to_date(date,'dd/MM/yyyy')) as Year,
        month(to_date(date,'dd/MM/yyyy')) as Month,
        day(to_date(date,'dd/MM/yyyy')) as day,
        quarter(to_date(date,'dd/MM/yyyy')) as Quarter,
        Region_Name ,
        Area_Code ,
        Cash_Average_Price ,
        Cash_Index ,
        coalesce(Cash_Monthly_Change ,0) as Cash_Monthly_Change,
        coalesce(Cash_Annual_Change ,0) as Cash_Annual_Change,
        Cash_Sales_Volume ,
        Mortgage_Average_Price ,
        Mortgage_Index ,
        coalesce(Mortgage_Monthly_Change ,0) as Mortgage_Monthly_Change,
        coalesce(Mortgage_Annual_Change ,0) as Mortgage_Annual_Change,
        Mortgage_Sales_Volume ,
        _rescued_data,
        date(current_timestamp()) as process_date
FROM  datamodelling.bronze.ukhousing 
"""
).createOrReplaceTempView("Silver_ukhousingmarket")
```

```python

spark.sql(
""" 
CREATE OR REPLACE TABLE datamodelling.silver.ukhousing
AS (    
    SELECT * FROM silver_ukhousingmarket
    )
"""
)
```

```python
%sql
select *  from datamodelling.silver.ukhousing
```

```python
%sql
MERGE INTO datamodelling.silver.ukhousing as target 
USING Silver_ukhousingmarket as source 
ON target.Date = source.Date AND 
   target.region_name=source.region_name  AND
   target.area_code=source.area_code
   
 WHEN MATCHED  THEN UPDATE SET *
 WHEN NOT MATCHED THEN INSERT * 

```
